##OPERADORES
###+ -> Suma

###- -> Resta

###/ -> Dividir

###* -> Multiplicar

###** -> Potencia

###> -> Mayor que

###< -> Menor que

###== -> Igual que


```python
a = 25.6
b = 52.90

#Operadores con número: suma
print("Suma:",a + b)
print (a + b)

#Otra manera, agregar:
c = a + b
print(c)

#Promedio
print("Promedio de dos lecturas", (a + b)/2)

#Comparaciones: True / False
print("¿a es menor que b?", a < b)
print("¿a es igual que b?", a == b)
```

```python
#EJERCICIO
valor_1 = 20.1
valor_2 = 19.8

#promedio
print("Promedio", (valor_1 + valor_2)/2)
```

##LISTAS


```python
#Expresiones: entre corchetes

expresiones = [12.5, 39, 46.7, 21]

#Posición indicada con índices: SIEMPRE SE INICIA CON 0 = ÍNDICE CERO
print("Segundo", expresiones[1])
print("Primero", expresiones[0])

#Conteo de cuantos datos tenemos
print("Datos guardados", len(expresiones))
len(expresiones)
```

```python
#Para agregar más datos/valores a nuestra lista
expresiones.append(13.1)
print("Lista actualizada", expresiones)
```

#DICCIONARIOS
Las {} indican los valores a tomar en cuenta

```python
H4PXi = {"localidad": "Cerro del Xihuingo,Tepeapulco, Hgo",
         "hospedero": "P. difficilis",
         "hembras": 20
}

print(H4PXi["hospedero"])
```

#BUCLE

Lectura en ciclo de los valores y los imprime en lista

```python
lecturas = [25, 12, 56.7, 39, 21]
for valor in lecturas:
  print(valor) #La sangría aquí indica que esta instrucción se repite
```
