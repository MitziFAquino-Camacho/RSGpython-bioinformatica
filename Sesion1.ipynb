##OPERADORES
###+ -> Suma

###- -> Resta

###/ -> Dividir

###* -> Multiplicar

###** -> Potencia

###> -> Mayor que

###< -> Menor que

###== -> Igual que


```python
a = 25.6
b = 52.90

#Operadores con número: suma
print("Suma:",a + b)
print (a + b)

#Otra manera, agregar:
c = a + b
print(c)

#Promedio
print("Promedio de dos lecturas", (a + b)/2)

#Comparaciones: True / False
print("¿a es menor que b?", a < b)
print("¿a es igual que b?", a == b)
```
